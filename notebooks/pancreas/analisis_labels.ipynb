```python
from pathlib import Path

PANCREAS_DIR = Path("/mnt/d/Universidad/analitica/proyecto_analitica2/data/source/pancreas")
LABELS_DIR = PANCREAS_DIR / "panorama_labels"

print("Contenido útil de panorama_labels:")
for p in sorted(LABELS_DIR.iterdir()):
    if p.name != ".git":
        print(" -", p.name, "|", "DIR" if p.is_dir() else "FILE")
```

```python
readme_candidates = [
    LABELS_DIR / "README.md",
    LABELS_DIR / "readme.md",
    LABELS_DIR / "README.txt",
]

readme_path = None
for p in readme_candidates:
    if p.exists():
        readme_path = p
        break

print("README encontrado:", readme_path)

if readme_path is not None:
    with open(readme_path, "r", encoding="utf-8", errors="ignore") as f:
        text = f.read()
    print(text[:5000])
else:
    print("No se encontró README en panorama_labels")
```

```python
auto_labels_dir = LABELS_DIR / "automatic_labels"

label_files = sorted(auto_labels_dir.glob("*.nii.gz")) if auto_labels_dir.exists() else []
image_files = sorted(PANCREAS_DIR.glob("*.nii.gz"))

print("Cantidad de imágenes:", len(image_files))
print("Cantidad de labels automáticas:", len(label_files))

print("\nPrimeras 20 labels:")
for p in label_files[:20]:
    print(" -", p.name)
```

```python
image_case_ids = {
    p.name.replace("_0000.nii.gz", "")
    for p in image_files
}

label_case_ids = {
    p.name.replace(".nii.gz", "")
    for p in label_files
}

matched = image_case_ids & label_case_ids
missing_labels = image_case_ids - label_case_ids
orphan_labels = label_case_ids - image_case_ids

print("Casos imagen:", len(image_case_ids))
print("Casos con label emparejada:", len(matched))
print("Imágenes sin label:", len(missing_labels))
print("Labels sin imagen:", len(orphan_labels))

print("\nPrimeras 20 imágenes sin label:")
for x in sorted(list(missing_labels))[:20]:
    print(" -", x)

print("\nPrimeras 20 labels sin imagen:")
for x in sorted(list(orphan_labels))[:20]:
    print(" -", x)
```

```python
import SimpleITK as sitk
import numpy as np

sample_label_path = label_files[0]
label_img = sitk.ReadImage(str(sample_label_path))
label_arr = sitk.GetArrayFromImage(label_img)

print("Sample label:", sample_label_path.name)
print("Shape:", label_arr.shape)
print("Spacing:", label_img.GetSpacing())
print("Unique values (primeros):", np.unique(label_arr)[:20])
print("Min/Max:", label_arr.min(), label_arr.max())
print("Suma total voxels:", label_arr.sum())
print("Voxels no cero:", int((label_arr != 0).sum()))
```

```python
sample_case = sample_label_path.name.replace(".nii.gz", "")
sample_image_path = PANCREAS_DIR / f"{sample_case}_0000.nii.gz"

print("Case:", sample_case)
print("Imagen existe:", sample_image_path.exists())
print("Label existe:", sample_label_path.exists())

if sample_image_path.exists():
    image_img = sitk.ReadImage(str(sample_image_path))
    image_arr = sitk.GetArrayFromImage(image_img)

    print("\nImagen:")
    print("  shape:", image_arr.shape)
    print("  spacing:", image_img.GetSpacing())
    print("  min/max:", float(image_arr.min()), float(image_arr.max()))

    print("\nLabel:")
    print("  shape:", label_arr.shape)
    print("  spacing:", label_img.GetSpacing())
    print("  voxels no cero:", int((label_arr != 0).sum()))
```

```python
from pathlib import Path
import pandas as pd

PANCREAS_DIR = Path("/mnt/d/Universidad/analitica/proyecto_analitica2/data/source/pancreas")
LABELS_DIR = PANCREAS_DIR / "panorama_labels"

clinical_path = LABELS_DIR / "clinical_information.xlsx"
print("clinical_path existe:", clinical_path.exists())

xls = pd.ExcelFile(clinical_path)
print("Hojas disponibles:", xls.sheet_names)
```

```python
df_clin = pd.read_excel(clinical_path, sheet_name=0)

print("Shape clinical:", df_clin.shape)
print("\nColumnas:")
print(df_clin.columns.tolist())

display(df_clin.head(10))
```

```python
for col in df_clin.columns:
    col_lower = str(col).lower()
    if any(key in col_lower for key in ["id", "case", "patient", "scan", "exam", "pdac", "diagn", "label", "reference"]):
        print(col)
```

```python
candidate_cols = [
    col for col in df_clin.columns
    if any(key in str(col).lower() for key in ["pdac", "diagn", "label", "reference", "case", "patient", "exam", "id"])
]

for col in candidate_cols:
    print("\n" + "="*80)
    print("Columna:", col)
    try:
        print(df_clin[col].value_counts(dropna=False).head(20))
    except Exception as e:
        print("No se pudo resumir:", e)
```
