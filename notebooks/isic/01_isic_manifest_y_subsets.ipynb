## imports 

```python
from pathlib import Path
import pandas as pd
import numpy as np

def find_project_root(start: Path) -> Path:
    current = start.resolve()
    for candidate in [current] + list(current.parents):
        if (candidate / "data").exists() and (candidate / "notebooks").exists():
            return candidate
    raise FileNotFoundError("No pude encontrar la raíz del proyecto.")

PROJECT_ROOT = find_project_root(Path.cwd())
DATA_DIR = PROJECT_ROOT / "data"
SOURCE_DIR = DATA_DIR / "source"
WORKING_DIR = DATA_DIR / "working"
MANIFESTS_DIR = DATA_DIR / "manifests"

MANIFESTS_DIR.mkdir(parents=True, exist_ok=True)
(WORKING_DIR / "isic" / "meta").mkdir(parents=True, exist_ok=True)
(WORKING_DIR / "isic" / "subsets").mkdir(parents=True, exist_ok=True)

print("PROJECT_ROOT:", PROJECT_ROOT)
print("DATA_DIR:", DATA_DIR)
print("SOURCE_DIR:", SOURCE_DIR)
print("WORKING_DIR:", WORKING_DIR)
print("MANIFESTS_DIR:", MANIFESTS_DIR)
```

## rutas base de ISIC

```python
isic_source_dir = SOURCE_DIR / "isic"

groundtruth_csv = isic_source_dir / "ISIC_2019_Training_GroundTruth.csv"
metadata_csv = isic_source_dir / "ISIC_2019_Training_Metadata.csv"

print("isic_source_dir:", isic_source_dir)
print("groundtruth_csv:", groundtruth_csv)
print("metadata_csv:", metadata_csv)

print("\n¿Existen?")
print("ISIC_2019_Training_GroundTruth.csv ->", groundtruth_csv.exists())
print("ISIC_2019_Training_Metadata.csv ->", metadata_csv.exists())
```

```python
print("Primer nivel dentro de data/source/isic:\n")
for p in sorted(isic_source_dir.iterdir()):
    print("-", p)
```

## cargar ground truth y metadata

```python
gt_df = pd.read_csv(groundtruth_csv)
meta_df = pd.read_csv(metadata_csv)

print("GroundTruth shape:", gt_df.shape)
print("Metadata shape:", meta_df.shape)

print("\nColumnas GroundTruth:")
print(gt_df.columns.tolist())

print("\nColumnas Metadata:")
print(meta_df.columns.tolist())

gt_df.head()
```

## unir ground truth y metadata

```python
df = gt_df.merge(meta_df, on="image", how="left")

print("Shape combinado:", df.shape)
df.head()
```

## definir columnas de clase y construir etiqueta principal

```python
class_columns = ["MEL", "NV", "BCC", "AK", "BKL", "DF", "VASC", "SCC", "UNK"]

def get_active_labels(row):
    labels = [col for col in class_columns if row[col] == 1.0]
    return labels

df["labels_list"] = df.apply(get_active_labels, axis=1)
df["label_count"] = df["labels_list"].apply(len)
df["target_label"] = df["labels_list"].apply(lambda x: x[0] if len(x) > 0 else "UNKNOWN")

print("Distribución de cantidad de labels activas:")
print(df["label_count"].value_counts().sort_index())

df[["image", "labels_list", "target_label"]].head()
```

## columnas base del proyecto

```python
df["dataset_name"] = "ISIC_2019"
df["modality"] = "dermoscopy"
df["dimension"] = "2D"

df.head()
```

## imágenes reales recursivamente

```python
image_files = list(isic_source_dir.rglob("*.jpg")) + list(isic_source_dir.rglob("*.jpeg")) + list(isic_source_dir.rglob("*.png"))

print("Cantidad total de imágenes encontradas:", len(image_files))
print("\nPrimeras 10:")
for p in image_files[:10]:
    print(p)
```

## mapa image -> file_path

```python
image_map = {p.stem: str(p.resolve()) for p in image_files}

print("Cantidad de entradas en image_map:", len(image_map))

sample_keys = list(image_map.keys())[:5]
print("\nEjemplos:")
for k in sample_keys:
    print(k, "->", image_map[k])
```

## Rutas del dataframe 

```python
df["file_path"] = df["image"].map(image_map)

missing_paths = df["file_path"].isna().sum()

print("Imágenes sin ruta encontrada:", missing_paths)
df[["image", "file_path"]].head()
```

## Renombre y limpieza de columnas

```python
df = df.rename(columns={
    "image": "image_name",
    "age_approx": "age_approx",
    "anatom_site_general": "anatom_site_general",
    "lesion_id": "lesion_id",
    "sex": "sex",
})

final_columns = [
    "image_name",
    "file_path",
    "target_label",
    "labels_list",
    "label_count",
    "age_approx",
    "anatom_site_general",
    "lesion_id",
    "sex",
    "dataset_name",
    "modality",
    "dimension",
] + class_columns

df = df[final_columns]

print("Columnas finales:")
print(df.columns.tolist())
df.head()
```

## split train / val / test

```python
RANDOM_SEED = 42
VAL_RATIO = 0.15
TEST_RATIO = 0.15

df = df.copy()

group_id = df["lesion_id"].fillna(df["image_name"]).astype(str)
unique_groups = group_id.unique()

rng = np.random.default_rng(RANDOM_SEED)
shuffled_groups = rng.permutation(unique_groups)

n_total = len(shuffled_groups)
n_test = int(n_total * TEST_RATIO)
n_val = int(n_total * VAL_RATIO)

test_groups = set(shuffled_groups[:n_test])
val_groups = set(shuffled_groups[n_test:n_test + n_val])
train_groups = set(shuffled_groups[n_test + n_val:])

df["group_id"] = group_id
df["split_final"] = "unknown"

df.loc[df["group_id"].isin(train_groups), "split_final"] = "train"
df.loc[df["group_id"].isin(val_groups), "split_final"] = "val"
df.loc[df["group_id"].isin(test_groups), "split_final"] = "test"

print("Conteo por split:")
print(df["split_final"].value_counts())

print("\nGrupos únicos por split:")
print("train:", df[df["split_final"] == "train"]["group_id"].nunique())
print("val:", df[df["split_final"] == "val"]["group_id"].nunique())
print("test:", df[df["split_final"] == "test"]["group_id"].nunique())
```

## Guardar manifiesto

```python
manifest_master_path = MANIFESTS_DIR / "manifest_isic_master.csv"
df.to_csv(manifest_master_path, index=False)

print("Manifest maestro guardado en:", manifest_master_path)
print("Shape final:", df.shape)
```

## Subset small 

```python
SMALL_TRAIN = 500
SMALL_VAL = 200
SMALL_TEST = 200
RANDOM_SEED = 42

train_small = df[df["split_final"] == "train"].sample(
    n=min(SMALL_TRAIN, len(df[df["split_final"] == "train"])),
    random_state=RANDOM_SEED
).copy()

val_small = df[df["split_final"] == "val"].sample(
    n=min(SMALL_VAL, len(df[df["split_final"] == "val"])),
    random_state=RANDOM_SEED
).copy()

test_small = df[df["split_final"] == "test"].sample(
    n=min(SMALL_TEST, len(df[df["split_final"] == "test"])),
    random_state=RANDOM_SEED
).copy()

isic_subset_small = pd.concat([train_small, val_small, test_small], ignore_index=True)

print("Conteo subset small:")
print(isic_subset_small["split_final"].value_counts())
print("Shape:", isic_subset_small.shape)
```

```python
subset_dir = WORKING_DIR / "isic" / "subsets"
subset_dir.mkdir(parents=True, exist_ok=True)

isic_subset_small_path = subset_dir / "isic_subset_small.csv"
isic_subset_small.to_csv(isic_subset_small_path, index=False)

print("Subset small guardado en:", isic_subset_small_path)
```

## Subset Large 

```python
LARGE_TRAIN = 10000
LARGE_VAL = 1500
LARGE_TEST = 1500
RANDOM_SEED = 42

train_large = df[df["split_final"] == "train"].sample(
    n=min(LARGE_TRAIN, len(df[df["split_final"] == "train"])),
    random_state=RANDOM_SEED
).copy()

val_large = df[df["split_final"] == "val"].sample(
    n=min(LARGE_VAL, len(df[df["split_final"] == "val"])),
    random_state=RANDOM_SEED
).copy()

test_large = df[df["split_final"] == "test"].sample(
    n=min(LARGE_TEST, len(df[df["split_final"] == "test"])),
    random_state=RANDOM_SEED
).copy()

isic_subset_large = pd.concat([train_large, val_large, test_large], ignore_index=True)

print("Conteo subset large:")
print(isic_subset_large["split_final"].value_counts())
print("Shape:", isic_subset_large.shape)
```

```python
isic_subset_large_path = subset_dir / "isic_subset_large.csv"
isic_subset_large.to_csv(isic_subset_large_path, index=False)

print("Subset large guardado en:", isic_subset_large_path)
```

```python
print("Archivos generados:")
print("-", manifest_master_path)
print("-", isic_subset_small_path)
print("-", isic_subset_large_path)

print("\n¿Existen?")
print("manifest_isic_master.csv ->", manifest_master_path.exists())
print("isic_subset_small.csv ->", isic_subset_small_path.exists())
print("isic_subset_large.csv ->", isic_subset_large_path.exists())
```
